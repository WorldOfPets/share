## 1. Списки
Список — ряд элементов под номерами. Нумерация с нуля.

```
индекс:     0        1        2        3
         +--------+--------+--------+--------+
список:  |   10   |   20   |   30   |   40   |
         +--------+--------+--------+--------+
индекс:    -4       -3       -2       -1
```

| Хочу | Пишу | Получаю |
|---|---|---|
| Создать | `a = [10, 20, 30, 40]` | `[10, 20, 30, 40]` |
| Первый элемент | `a[0]` | `10` |
| Последний | `a[-1]` | `40` |
| Заменить | `a[0] = 7` | `[7, 20, 30, 40]` |

| Удалить элемент | `a.remove(50)` | список без 50 |
| Сколько элементов | `len(a)` | число |
| Сумма | `sum(a)` | число |


````python
# Пример
ocenki = [4, 5, 3, 5]
print("первая:", ocenki[0], "| последняя:", ocenki[-1])
ocenki.append(4)
print("после append:", ocenki)
print("всего:", len(ocenki), "| сумма:", sum(ocenki))
print("средняя:", sum(ocenki) / len(ocenki))
````

## 2. Кортежи
Кортеж — тот же список, но ЗАПЕЧАТАННЫЙ: элементы менять нельзя.

| | Список | Кортеж |
|---|---|---|
| Скобки | `[1, 2]` | `(1, 2)` |
| Менять элементы | можно | нельзя (ошибка) |
| Когда брать | данные меняются | данные постоянны: точки, даты, строки таблиц |

| Хочу | Пишу | Получаю |
|---|---|---|
| Создать | `t = (3, 5)` | `(3, 5)` |
| Один элемент | `t = (7,)` | запятая обязательна |
| Элемент | `t[0]` | `3` |
| Распаковать | `x, y = t` | x = 3, y = 5 |
| Сколько раз встречается | `t.count(5)` | `1` |
| Номер элемента | `t.index(5)` | `1` |


````python
# Пример
tochka = (3, 5)
x, y = tochka          # распаковка: два имени = два значения
print("x =", x, "| y =", y)
print("в кортеже", len(tochka), "элемента, менять их нельзя")
````

## 4. Словари
Словарь — справочник «ключ: значение», как телефонная книга.

```
slovar = {"Anna":  "555-01-01",
          "Boris": "555-01-02"}
           КЛЮЧ      ЗНАЧЕНИЕ
```

| Хочу | Пишу | Получаю |
|---|---|---|
| Создать | `d = {"a": 1}` | `{"a": 1}` |
| Взять по ключу | `d["a"]` | `1` |
| Взять БЕЗ ошибки | `d.get("z", 0)` | `0`, если ключа нет |
| Добавить / заменить | `d["b"] = 2` | новая пара |
| Удалить | `del d["a"]` | пары нет |
| Все ключи | `list(d.keys())` | список ключей |
| Все значения | `list(d.values())` | список значений |
| Все пары | `list(d.items())` | список кортежей |


````python
# Пример
telefon = {"Anna": "555-01-01", "Boris": "555-01-02"}
print(telefon["Anna"])
telefon["Vera"] = "555-01-03"        # добавили пару
for imya, nomer in telefon.items():  # пары распаковываются
    print(imya, "->", nomer)
````

## 5. Множества
Множество — мешок, где каждый элемент живёт ОДИН раз, порядок не важен.

| Хочу | Пишу | Получаю |
|---|---|---|
| Создать | `s = {1, 2, 2, 3}` | `{1, 2, 3}` — дубль исчез |
| Из списка | `set([1, 1, 2])` | `{1, 2}` |
| Добавить | `s.add(4)` | |
| Удалить | `s.remove(4)` | ошибка, если нет |
| Удалить мягко | `s.discard(4)` | без ошибки |
| Объединение | `a | b` | всё из обоих |
| Пересечение | `a & b` | общее |
| Разность | `a - b` | из a без b |
| Только в одном | `a ^ b` | без общих |
| Есть ли? | `2 in s` | True / False |


````python
# Пример
zayavki = ["Anna", "Boris", "Anna", "Vera"]
print(set(zayavki))                 # дубликаты исчезли
a = {1, 2, 3}
b = {2, 3, 4}
print("объединение:", a | b)
print("общее:", a & b)
print("только в a:", a - b)
````

## 6. List comprehension
Весь цикл — в одну строку. Читай вслух: «список из ВЫРАЖЕНИЕ для каждого ЭЛЕМЕНТ из ОТКУДА».

| Обычный цикл | То же в одну строку |
|---|---|
| `res = []`<br>`for x in range(5):`<br>`    res.append(x * x)` | `res = [x * x for x in range(5)]` |
| то же + условие<br>`    if x % 2 == 0:` | `res = [x * x for x in range(5) if x % 2 == 0]` |

| Часть строки | За что отвечает |
|---|---|
| `x * x` | что кладём в новый список |
| `for x in range(5)` | откуда берём |
| `if x % 2 == 0` | фильтр (необязательно) |


````python
# Пример
kvadraty = [x * x for x in range(1, 6)]
print(kvadraty)                              # [1, 4, 9, 16, 25]
chetnye = [x for x in range(10) if x % 2 == 0]
print(chetnye)                               # [0, 2, 4, 6, 8]
````

## 7. Упаковка и распаковка
Сколько имён слева — столько значений справа. Звёздочка `*` забирает «всё остальное» списком.

| Пишу | Получаю |
|---|---|
| `a, b = 1, 2` | a = 1, b = 2 |
| `a, b = b, a` | значения поменялись |
| `pervyy, *ost = [1, 2, 3, 4]` | pervyy = 1, ost = `[2, 3, 4]` |
| `*nachalo, posledniy = [1, 2, 3, 4]` | nachalo = `[1, 2, 3]`, posledniy = 4 |

## 8. Упаковка и распаковка в параметрах функции
`*args` собирает позиционные аргументы в кортеж. `**kwargs` собирает именованные — в словарь.

| Где | Пишу | Внутри функции |
|---|---|---|
| В параметрах | `def f(*args):` | `args` — кортеж всех позиционных |
| В параметрах | `def f(**kwargs):` | `kwargs` — словарь именованных |
| При вызове | `f(*spisok)` | список разложится по позициям |
| При вызове | `f(**slovar)` | словарь разложится по именам |


````python
# Пример
def summa(*chisla):
    return sum(chisla)

def anketa(**polia):
    for klyuch, znachenie in polia.items():
        print(klyuch, ":", znachenie)

print(summa(1, 2, 3))        # 6
print(summa(*[10, 20]))      # 30: список распаковался
anketa(imya="Anna", god=2009)
````

## 9. Функция zip
`zip` — молния-застёжка: соединяет элементы нескольких последовательностей в кортежи-пары.

```
imena:    Anna    Boris    Vera          zip(imena, ocenki)
ocenki:    5        3        4      ->   (Anna,5) (Boris,3) (Vera,4)
```

| Хочу | Пишу | Получаю |
|---|---|---|
| Пары | `list(zip(a, b))` | `[(a0, b0), (a1, b1), ...]` |
| Словарь | `dict(zip(klyuchi, znacheniya))` | `{k0: v0, ...}` |
| Цикл по двум спискам | `for x, y in zip(a, b):` | параллельный обход |
| Транспонировать матрицу | `list(zip(*m))` | строки становятся столбцами |

Важно: `zip` остановится на самой КОРОТКОЙ последовательности и выдаёт результат лениво — чтобы увидеть, оберни в `list()`.


````python
# Пример
imena = ["Anna", "Boris", "Vera"]
ocenki = [5, 3, 4]
for imya, ocenka in zip(imena, ocenki):
    print(imya, "получил", ocenka)
print(dict(zip(imena, ocenki)))
````

## Шпаргалка по части 1

| Тема | Главная мысль | Пример |
|---|---|---|
| Списки | ряд с номерами с нуля, можно менять | `a[0]`, `a.append(x)` |
| Кортежи | запечатанный список | `x, y = (3, 5)` |
| Диапазоны | счётчик, конец не входит | `range(1, 6)` |
| Словари | ключ: значение | `d.get(k, 0)` |
| Множества | без дублей, операции `|` `&` `-` `^` | `set(spisok)` |
| List comprehension | цикл в одну строку | `[x*x for x in a if x > 0]` |
| Распаковка | имён слева = значений справа, `*` — остаток | `a, *ost = spisok` |
| `*args`, `**kwargs` | упаковка в вызове и в параметрах | `f(*spisok)`, `f(**slovar)` |
| zip | параллельный обход, конец по короткому | `dict(zip(a, b))` |
